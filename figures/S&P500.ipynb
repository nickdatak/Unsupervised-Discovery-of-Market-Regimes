```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('/Users/nick/PycharmProjects/Unsupervised-Discovery-of-Market-Regimes/data/S&P500.csv')
df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
```

```python
print(df.head())
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["Close"], label="S&P 500 Close", color="tab:blue")
plt.xlabel("Date")
plt.ylabel("Index level")
plt.title("S&P 500 Closing Price Over Time")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

#Printing the whole chart
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["Open"], label="Open", alpha=0.7)
plt.plot(df.index, df["High"], label="High", alpha=0.7)
plt.plot(df.index, df["Low"], label="Low", alpha=0.7)
plt.plot(df.index, df["Close"], label="Close", alpha=0.7)
plt.xlabel("Date")
plt.ylabel("Index level")
plt.title("S&P 500 OHLC")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

#Printing Open/High/Low/Close for the whole chart
```

```python
period = df.loc["2000-01-01":"2001-01-30"]
start = period.index[0]
end = period.index[-1]

#To change that for a specific period
fig, ax_price = plt.subplots(figsize=(12, 6))

ax_price.plot(period.index, period["Close"], color="tab:blue", label="Close")
ax_price.set_xlabel("Date")
ax_price.set_ylabel("S&P 500 level", color="tab:blue")
ax_price.tick_params(axis="y", labelcolor="tab:blue")

ax_vol = ax_price.twinx()
ax_vol.bar(period.index, period["Volume"], color="tab:gray", alpha=0.3, width=1.0, label="Volume")
ax_vol.set_ylabel("Volume", color="tab:gray")
ax_vol.tick_params(axis="y", labelcolor="tab:gray")

plt.title("S&P 500 Close and Volume ({0} to {1})".format(start, end))
fig.tight_layout()
plt.show()

#Displaying prices and the volume on the same chart
```
