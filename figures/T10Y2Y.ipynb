```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('/data/t10y2y.csv')
df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
print(df.head())
```

```python
df = df.replace("", np.nan)
df = df.dropna(how="any")
print(df)

#Removing non-trading days from the dataset
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["T10Y2Y"], label="T10Y2Y", color="tab:blue")
plt.xlabel("Date")
plt.ylabel("T10Y2Y")
plt.title("T10Y2Y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
period = df.loc["2005-01-01":"2010-12-31"]
start = period.index[0]
end = period.index[-1]

#To change that for a specific period
fig, ax_price = plt.subplots(figsize=(12, 6))

ax_price.plot(period.index, period["T10Y2Y"], color="tab:blue", label="Close")
ax_price.set_xlabel("Date")
ax_price.set_ylabel("T10Y2Y", color="tab:blue")
ax_price.tick_params(axis="y", labelcolor="tab:blue")

plt.title("T10Y2Y value ({0} to {1})".format(start, end))
fig.tight_layout()
plt.show()
```
