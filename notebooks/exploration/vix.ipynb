```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sys.path.insert(0, str(Path.cwd().parent.parent))
from regime_utils import read_csv
```

```python
df = read_csv("vix.csv")
df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
```

```python
print(df.head())
```

```python
df.drop('Volume', axis=1, inplace=True)
print(df.head())
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["Close"], label="VIX Close", color="tab:blue")
plt.xlabel("Date")
plt.ylabel("Index level")
plt.title("VIX Closing Index")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["Open"], label="Open", alpha=0.7)
plt.plot(df.index, df["High"], label="High", alpha=0.7)
plt.plot(df.index, df["Low"], label="Low", alpha=0.7)
plt.plot(df.index, df["Close"], label="Close", alpha=0.7)
plt.xlabel("Date")
plt.ylabel("Index level")
plt.title("VIX OHLC")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
period = df.loc["2005-01-01":"2010-12-31"]
start = period.index[0]
end = period.index[-1]

#To change that for a specific period
fig, ax_price = plt.subplots(figsize=(12, 6))

ax_price.plot(period.index, period["Close"], color="tab:blue", label="Close")
ax_price.set_xlabel("Date")
ax_price.set_ylabel("VIX", color="tab:blue")
ax_price.tick_params(axis="y", labelcolor="tab:blue")

plt.title("VIX value ({0} to {1})".format(start, end))
fig.tight_layout()
plt.show()
```
