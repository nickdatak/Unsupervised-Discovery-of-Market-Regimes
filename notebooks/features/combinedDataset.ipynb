```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

sys.path.insert(0, str(Path.cwd().parent.parent))
from regime_utils import read_csv, write_csv
```

```python
sp500 = read_csv("sp500_weekly.csv", parse_dates=["Date"])
vix = read_csv("vix_weekly.csv", parse_dates=["Date"])
yields = read_csv("t10_weekly.csv", parse_dates=["Date"])
```

```python
sp500 = sp500.set_index("Date")
vix = vix.set_index("Date")
yields = yields.set_index("Date")
```

```python
sp500_features = sp500[["Adj Close"]].rename(
    columns={"Adj Close": "SP500"}
)
sp500_features["SP500_Return"] = np.log(sp500_features["SP500"]).diff()
#Computing returns after weekly aggregation
```

```python
vix_features = vix.rename(columns={"Adj Close": "VIX"})
```

```python
yield_features = yields.rename(columns={"T10Y2Y": "Yield_Spread"})
```

```python
features = (
    sp500_features
    .join(vix_features, how="inner")
    .join(yield_features, how="inner")
)
```

```python
print(features.head())
```

```python
features.dropna(inplace=True)
print(features.head())
```

```python
features = features.drop(columns=['Close', 'High', 'Low', 'Open'])
```

```python
#Sanity checks
print(features.head())
print(features.tail())
print(features.isna().sum())
print(features.index.min(), features.index.max())
```

```python
write_csv(features, "market_features_weekly.csv")
#creating a dataset for model training
```

```python
plt.figure(figsize=(12, 6))
plt.plot(features.index, features["SP500"], label="SP500", alpha=0.7)
plt.plot(features.index, features["SP500_Return"], label="SP500_Return", alpha=0.7)
plt.plot(features.index, features["VIX"], label="VIX", alpha=0.7)
plt.plot(features.index, features["Yield_Spread"], label="Yield_Spread", alpha=0.7)
plt.xlabel("Date")
plt.ylabel("Feature level")
plt.title("Plotted Features")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
