# 08 — VIX standardization comparison

Overlay standardized VIX from three schemes:

- **Full-sample (baseline):** committed `market_features_weekly_std.csv`
- **Rolling (52w):** causal z-score with a 52-week window
- **Expanding:** causal z-score with expanding mean/std

Plot uses the common date range where rolling and expanding are defined (first 51 weeks omitted).

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

sys.path.insert(0, str(Path.cwd().parent))
from regime_utils import *

full_vix = load_full_sample_std()["VIX"]
rolling_vix = load_rolling_std()["VIX"]
expanding_vix = load_expanding_std()["VIX"]

common_idx = full_vix.index.intersection(rolling_vix.index).intersection(expanding_vix.index)
print(f"Full-sample: {len(full_vix)} weeks")
print(f"Rolling:     {len(rolling_vix)} weeks")
print(f"Expanding:   {len(expanding_vix)} weeks")
print(f"Overlay:     {len(common_idx)} weeks ({common_idx.min().date()} → {common_idx.max().date()})")
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(common_idx, full_vix.loc[common_idx], label="Full-sample", alpha=0.85, linewidth=1.2)
ax.plot(common_idx, rolling_vix.loc[common_idx], label="Rolling (52w)", alpha=0.85, linewidth=1.2)
ax.plot(common_idx, expanding_vix.loc[common_idx], label="Expanding", alpha=0.85, linewidth=1.2)

ax.axhline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.4)
ax.set_title("Standardized VIX — full-sample vs rolling vs expanding")
ax.set_xlabel("Date")
ax.set_ylabel("Z-score")
ax.legend(loc="upper right")
ax.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```
